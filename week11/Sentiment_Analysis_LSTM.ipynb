## Sentiment Analysis

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Embedding
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python

class LSTMClassifier:
    
    def __init__(self, X_train, y_train, X_test, y_test, max_words=2000, max_len=100, embedding_dim=50, lstm_units=32, num_classes=2):
        self.X_train = X_train
        self.X_test = X_test
        self.y_train = y_train
        self.y_test = y_test
        self.max_words = max_words
        self.max_len = max_len
        self.embedding_dim = embedding_dim
        self.lstm_units = lstm_units
        self.num_classes = num_classes
        
        self.tokenizer = None
        self.model = None
    
    def preprocess(self, X_train, y_train, X_test, y_test):
        self.tokenizer = Tokenizer(num_words=self.max_words)
        self.tokenizer.fit_on_texts(X_train)

        X_train_seq = self.tokenizer.texts_to_sequences(X_train)
        X_test_seq = self.tokenizer.texts_to_sequences(X_test)

        self.vocab_size = len(self.tokenizer.word_index) + 1

        X_train_pad = pad_sequences(X_train_seq, maxlen=self.max_len)
        X_test_pad = pad_sequences(X_test_seq, maxlen=self.max_len)

        self.X_train = np.array(X_train_pad)
        self.X_test = np.array(X_test_pad)
        self.y_train = np.array(y_train)
        self.y_test = np.array(y_test)
        
    def build_model(self):
        self.model = Sequential()
        self.model.add(Embedding(input_dim=self.max_words, output_dim=self.embedding_dim, input_length=self.max_len))
        self.model.add(LSTM(units=self.lstm_units))
        self.model.add(Dense(units=self.num_classes, activation='softmax'))
        self.model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
    
    def fit(self, epochs=10, batch_size=32):
        self.model.fit(self.X_train, self.y_train, validation_data=(self.X_test, self.y_test), epochs=epochs, batch_size=batch_size)
    
    def evaluate(self):
        loss, accuracy = self.model.evaluate(self.X_test, self.y_test)
        print(f'Test Loss: {loss:.3f}')
        print(f'Test Accuracy: {accuracy:.3f}')
    
    def predict(self, X):
        X_seq = self.tokenizer.texts_to_sequences(X)
        X_pad = pad_sequences(X_seq, maxlen=self.max_len)
        y_pred = self.model.predict(X_pad)
        return y_pred
```

```python
from keras.datasets import imdb

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=2000)

# Convert the integer sequences to strings
word_index = imdb.get_word_index()
reverse_word_index = dict([(value, key) for (key, value) in word_index.items()])
X_train = [' '.join([reverse_word_index.get(i - 3, '?') for i in x]) for x in X_train]
X_test = [' '.join([reverse_word_index.get(i - 3, '?') for i in x]) for x in X_test]

# Preprocess the data and train the model
lstm_classifier = LSTMClassifier(X_train, y_train, X_test, y_test)
lstm_classifier.preprocess(X_train, y_train, X_test, y_test)
lstm_classifier.build_model()
lstm_classifier.fit(epochs=10, batch_size=32)
lstm_classifier.evaluate()
```
